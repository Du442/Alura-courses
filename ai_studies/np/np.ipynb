```python
Y= Moscow
X = datas
n = np.size(Moscow)
```

```python
(X**2).shape
```

```python
(n*np.sum(X*Y) - np.sum(X)*np.sum(Y)) / (n*np.sum(np.power(X, 2)) - np.power(np.sum(X), 2))
```

```python
a = (n*np.sum(X*Y) - np.sum(X)*np.sum(Y))/(n*np.sum(X**2) - np.sum(X)**2)
```
