Atividade 1 -

```python
caminho_url = 'https://raw.githubusercontent.com/allanspadini/numpy/dados/citrus.csv'
cols = np.arange(1, 6, 1)

dado = np.loadtxt(caminho_url, delimiter=',', usecols=cols, skiprows=1)
```

```python
dado
```

Atividade 2 -

```python
dados_transpostos = dado.T
```

```python
dados = dados_transpostos
```

```python
diametro_laranja = dados[0,:5000]
diametro_toranja = dados[0,5000:]
peso_laranja = dados[1,:5000]
peso_toranja = dados[1,5000:]
```

```python
diametro_laranja
```

```python
diametro_laranja.shape
```

Atividade 3 - 

```python
Y = peso_toranja
X = diametro_toranja
n = np.size(X)

a = (n*sum(X*Y) - sum(X)*sum(Y)) / (n*sum(np.power(X,2)) - (sum(X))**2)
b = np.mean(Y) - a * np.mean(X)
```

```python
np.random.seed(84)
coef_angulares = np.random.uniform(0, 30.0, 100)

norma = np.array([])
for i in range(100):
    
    norma = np.append(norma, np.linalg.norm(Y - (coef_angulares[i]*X+b)))
```
