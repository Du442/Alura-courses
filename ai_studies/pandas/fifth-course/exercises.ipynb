```python
import pandas as pd
import json
```

```python
cursos = pd.read_json('cursos_cadastrados.json')
```

```python
with open('cursos_cadastrados.json', 'r', encoding='utf-8') as f:
    cursos = json.load(f)
```

```python
cursos = pd.json_normalize(cursos)
```

```python
cursos.info()
```

```python
cursos['carga_horaria'] = cursos['carga_horaria'].fillna('10 horas', inplace=True)
```

```python
cursos['descricao'] = cursos['descricao'].fillna('Curso introdutório para Excel.')
```

```python
cursos
```

```python
cursos.at[3, "instrutor.nome"] = 'Teste'
cursos
```

```python
cursos['preco'].isna()
cursos['preco'] = cursos['preco'].fillna(79.90)
cursos.head()
```

```python
cursos['concluintes'] = cursos['concluintes'].fillna(80)
cursos['data_inicio'] = cursos['data_inicio'].fillna('2022-03-22')
cursos['data_conclusao'] = cursos['data_conclusao'].fillna('2022-04-16')
```

```python
cursos
```

```python
cursos.head()
```

```python
cursos.at[3, "concluintes"] = 92
cursos.at[3, 'data_conclusao'] = '2022-04-26'
cursos.at[3, "instrutor.telefone"] = '(11) 6666-6666'

cursos.at[4, "data_inicio"] = '2022-07-18'
cursos.at[4, "data_conclusao"] = '2022-08-01'
cursos.at[4, "instrutor.email"] = 'testeteste@gmail.com'
cursos.at[4, "instrutor_telefone"] = '(11) 5555-5555'
```
