```python
import pandas as pd
```

```python
path = 'moveis_disponiveis.json'
path2 = 'dados_hospedagem.json'

moveis_disponiveis = pd.read_json(path)
dados_hospedagem = pd.read_json(path2)
```

```python
dados_hospedagem_normalizado = pd.json_normalize(dados_hospedagem['info_moveis'])
```

<ul>
    <li>avaliacao_geral: refere-se à média de notas dadas para a avaliação da hospedagem no imóvel.</li>
    <li>experiencia_local: descreve as experiências oferecidas durante a hospedagem no imóvel.</li>
    <li>max_hospedes: informa a quantidade máxima de hóspedes que o local permite.</li>
    <li>descricao_local: descreve o imóvel.</li>
    <li>descricao_vizinhanca: descreve a vizinhança ao redor do imóvel.</li>
    <li>quantidade_banheiros: informa a quantidade de banheiros disponíveis.</li>
    <li>quantidade_quartos: informa a quantidade de quartos disponíveis.</li>
    <li>quantidade_camas: informa a quantidade de camas disponíveis.</li>
    <li>modelo_cama: informa o modelo de cama oferecido.</li>
    <li>comodidades: informa as comodidades oferecidas pelo imóvel.</li>
    <li>taxa_deposito: informa a taxa de depósito mínima para segurança de hospedagem.</li>
    <li>taxa_limpeza: informa a taxa cobrada para o serviço de limpeza.</li>
    <li>preco: refere-se ao preço base a ser cobrado pela diária no imóvel.</li>
</ul>

```python
colunas = list(dados_hospedagem_normalizado.columns)
colunas
```

```python
dados_hospedagem_normalizado = dados_hospedagem_normalizado.explode(colunas[3:])
dados_hospedagem_normalizado
```

```python
dados_hospedagem_normalizado.reset_index(inplace=True, drop=True)
dados_hospedagem_normalizado
```

```python
dados_hospedagem_normalizado.info()
```

```python
import numpy as np
```

```python
dados_hospedagem_normalizado['max_hospedes'] = dados_hospedagem_normalizado['max_hospedes'].astype(int)
```

```python
dados_hospedagem_normalizado.info()
```

```python
col_numericas = ['quantidade_banheiros', 'quantidade_quartos', 'quantidade_camas']

dados_hospedagem_normalizado[col_numericas] = dados_hospedagem_normalizado[col_numericas].astype(int)
```

```python
dados_hospedagem_normalizado.info()
```

```python
dados_hospedagem_normalizado['avaliacao_geral'] = dados_hospedagem_normalizado['avaliacao_geral'].astype(float)
```

```python
dados_hospedagem_normalizado.info()
```

```python
dados_hospedagem_normalizado['preco'] = dados_hospedagem_normalizado['preco'].astype(float)
```

```python
dados_hospedagem_normalizado['preco'] = dados_hospedagem_normalizado['preco'].apply(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
dados_hospedagem_normalizado['preco']
```

```python
dados_hospedagem_normalizado['preco'] = dados_hospedagem_normalizado['preco'].astype(float)
```

```python
dados_hospedagem_normalizado.info()
```

```python
dados_hospedagem_normalizado[['taxa_deposito', 'taxa_limpeza']] = dados_hospedagem_normalizado[['taxa_deposito', 'taxa_limpeza']].map(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
dados_hospedagem_normalizado.info()
```

```python
dados_hospedagem_normalizado[['taxa_deposito', 'taxa_limpeza']] = dados_hospedagem_normalizado[['taxa_deposito', 'taxa_limpeza']].astype(float)
```

```python
dados_hospedagem_normalizado.info()
```

```python
dados_hospedagem_normalizado['descricao_local'] = dados_hospedagem_normalizado['descricao_local'].str.lower()
```

```python
dados_hospedagem_normalizado.head()
```

```python
dados_hospedagem_normalizado['descricao_local'] = dados_hospedagem_normalizado['descricao_local'].str.replace('[^a-zA-Z0-9\-\']', ' ', regex=True)
```

```python
dados_hospedagem_normalizado['descricao_local'] = dados_hospedagem_normalizado['descricao_local'].str.replace('(?<!\w)-(?!\w)', ' ', regex=True)
```

```python
dados_hospedagem_normalizado['descricao_local'] = dados_hospedagem_normalizado['descricao_local'].str.split()
dados_hospedagem_normalizado.head()
```

```python
dados_hospedagem_normalizado['comodidades'] = dados_hospedagem_normalizado['comodidades'].str.replace('\{|}|\"','',regex=True)
```

```python
dados_hospedagem_normalizado['comodidades'] = dados_hospedagem_normalizado['comodidades'].str.split(',')
```
