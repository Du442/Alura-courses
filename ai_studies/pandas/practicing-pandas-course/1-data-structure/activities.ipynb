Esses dados estão organizados de forma um pouco confusa, dificultando a análise. Sua tarefa é usar um loop for em Python para processar essa lista e organizar as informações de cada atleta em um formato mais estruturado, facilitando futuras análises estatísticas e comparações.<p>

```python
atletas = [
    ["Maria Silva", 1.75, 65],
    ["João Santos", 1.80, 72],
    ["Ana Pereira", 1.68, 58],
    ["Pedro Oliveira", 1.92, 85],
    ["Carlos Lima", 1.85, 78],
    ["Beatriz Souza", 1.70, 60],
    ["Fernanda Costa", 1.62, 55],
    ["Lucas Almeida", 1.88, 82],
    ["Rafaela Gomes", 1.74, 63],
    ["Gustavo Ferreira", 1.90, 88],
    ["Larissa Rocha", 1.66, 57],
    ["Henrique Nunes", 1.83, 76],
    ["Juliana Martins", 1.72, 59],
    ["Ricardo Carvalho", 1.86, 80],
    ["Sofia Alves", 1.64, 54],
    ["Matheus Ribeiro", 1.89, 84],
    ["Camila Duarte", 1.69, 61],
    ["Gabriel Monteiro", 1.77, 73],
    ["Eduarda Farias", 1.71, 62],
    ["Thiago Mendes", 1.84, 79],
]
```

<p>Objetivo:

Utilizando um loop for, crie um programa que processe a lista atletas e para cada atleta, imprima as informações no seguinte formato:</p><br>

<p>2. Continuando o seu trabalho como cientista de dados do comitê olímpico, agora você deve reorganizar a lista de atletas, em uma lista de tuplas, onde cada tupla representa um atleta e contém as informações de nome, altura e peso, respectivamente. Essa estrutura permitirá o acesso aos dados de cada atleta de forma mais eficiente, e que esses dados não possam mais ser modificados. Essa organização é crucial para etapas posteriores da análise, como o cálculo do IMC (Índice de Massa Corporal) de cada atleta e a identificação de padrões nos dados.

<strong>Objetivo:</strong>

Escreva um programa Python que converta a lista atletas em uma lista de tuplas, onde cada tupla contém o nome, altura e peso de um atleta. Imprima a lista de tuplas resultante.</p>

```python
lista_de_tuplas = []

for i in atletas:
    nome = i[0]
    altura = i[1]
    peso = i[2]
    atleta_tupla = (nome, altura, peso)
    lista_de_tuplas.append(atleta_tupla)

print(lista_de_tuplas)
```

<p>Agora, queremos identificar quais atletas possuem IMC (Índice de Massa Corporal) acima de 25, que é considerado como sobrepeso. O cálculo do IMC pode ser feito com a fórmula:

Utilizando list comprehension, crie uma lista contendo o nome dos atletas que possuem IMC maior que 25 e imprima essa lista.</p>

<p>4. Estamos trabalhando com o mesmo conjunto de dados contendo informações sobre atletas, já no formato de tuplas.

Queremos classificar os atletas com base em seu IMC (Índice de Massa Corporal) como "Peso Normal" ou "Sobrepeso". Porém, dessa vez queremos uma lista mais elaborada. O cálculo do IMC pode ser feito com a fórmula:

Fórmula do IMC: IMC = peso / altura².

Utilizando list comprehension, crie uma nova lista contendo tuplas no formato:

(nome_do_atleta, "Peso Normal" ou "Sobrepeso")
Copiar código
Se o IMC do atleta for maior que 25, ele será classificado como "Sobrepeso". Caso contrário, será classificado como "Peso Normal".

Imprima a lista resultante. Dica: Você pode usar if e else no list comprehension para realizar a tarefa.</p>

```python
classificacao_imc = [
    (atleta[0], "Sobrepeso" if (atleta[2] / (atleta[1] ** 2)) > 25 else "Peso Normal")
    for atleta in atletas
]
```

Explicação da Solução
1 - A lista atletas contém informações sobre os atletas no formato de tuplas, onde:

O índice 0 contém o nome do atleta.
O índice 1 contém a altura do atleta (em metros).
O índice 2 contém o peso do atleta (em kg).
2 - Utilizamos list comprehension para criar uma nova lista chamada classificacao_imc, que contém tuplas no formato (nome_do_atleta, classificacao):

classificacao_imc = [
    (atleta[0], "Sobrepeso" if (atleta[2] / (atleta[1] ** 2)) > 25 else "Peso Normal")
    for atleta in atletas
]
Copiar código
Para cada tupla atleta na lista atletas, verificamos o IMC do atleta usando a fórmula peso/altura^2, onde atleta[2] é o peso e atleta[1] é a altura.

Se o IMC do atleta for maior que 25, a classificação será "Sobrepeso". Caso contrário, será "Peso Normal".

O resultado de cada iteração é uma nova tupla contendo o nome do atleta (atleta[0]) e sua classificação.

3 - Por fim, imprimimos a lista de classificação.

```python
print("Classificação dos atletas:")
print(classificacao_imc)
```

<p>9. Você trabalha como cientista de dados em uma empresa que armazena informações sobre seus usuários. Foi solicitado que você crie uma classificação simples dos usuários com base na idade. A classificação deve ser feita em duas categorias:

"adulto": Para usuários com 18 anos ou mais.
"menor": Para usuários com menos de 18 anos.

Os dados dos usuários estão armazenados em um dicionário chamado usuarios, no seguinte formato:

```python
usuarios = {
    "João": 25,
    "Maria": 17,
    "Ana": 19,
    "Carlos": 16,
    "Beatriz": 22,
    "Pedro": 15,
    "Luiza": 18
}
```

Usando dict comprehension com if e else, crie um novo dicionário chamado classificacao_usuarios, onde:

As chaves serão os nomes dos usuários.

Os valores serão as strings "adulto" ou "menor", dependendo da idade.

O dicionário resultante deve ter o seguinte formato:

```python
classificacao_usuarios = {
    item: 'Maior'
    if info >= 18 else 'Menor'
    for item, info in usuarios.items()
}
```
