<p>7. Juan recebeu um conjunto de dados em formato JSON e precisa normalizá-lo para acessar as informações de maneira estruturada. No entanto, ao utilizar json_normalize, ele identificou que o arquivo continha duas listas

Como Juan pode estruturar esses dados em um único DataFrame?</p>

```python
with open('arquivos/questao_7.json', 'r', encoding='utf-8') as f:
    df_7 = json.load(f)
```

```python
df_7
```

```python
lojas_sudeste = pd.json_normalize(df_7, record_path='lojas_sudeste',)
lojas_sul = pd.json_normalize(df_7, record_path='lojas_sul')
```

```python
todas_lojas = pd.merge(lojas_sudeste, lojas_sul, 'outer', indicator=True)
```

```python
todas_lojas.reset_index(drop=True, inplace=True)
```

```python
todas_lojas
```

<p>Segunda alternativa</p>

```python
df_final = pd.concat([lojas_sudeste, lojas_sul], ignore_index=True)
df_final
```

<p>8. Paula está trabalhando com um conjunto de dados que contém duas listas aninhadas dentro de uma chave chamada lojas. Devido a essa estrutura, ela está enfrentando dificuldades para transformar o documento em um DataFrame manipulável.</p>

```python
pd.read_json('arquivos/questao_8.json')
```

```python
with open('arquivos/questao_8.json', 'r', encoding='utf-8') as f:
    df_8 = json.load(f)
```

```python
df_8
```

```python
df_8_norte = pd.json_normalize(df_8, record_path=['lojas', 'lojas_norte'])
df_8_nordeste = pd.json_normalize(data=df_8, record_path=['lojas', 'lojas_nordeste'])
```

```python
df_8 = pd.concat([df_8_nordeste, df_8_norte], ignore_index=True)
```

```python
df_8
```

```python
dados = [
    {
        "id_loja": 17,
        "nome_loja": "Loja Rho",
        "detalhes": [
            {"cidade": "João Pessoa", "estado": "PB"}
        ],
        "vendas": [
            {"categoria": "Roupas", "vendas_mensais": 125, "faturamento_mensal": 37500.0}
        ]
    },
    {
        "id_loja": 18,
        "nome_loja": "Loja Sigma",
        "detalhes": [
            {"cidade": "Vitória", "estado": "ES"}
        ],
        "vendas": [
            {"categoria": "Calçados", "vendas_mensais": 180, "faturamento_mensal": 54000.0}
        ]
    },
    {
        "id_loja": 19,
        "nome_loja": "Loja Tau",
        "detalhes": [
            {"cidade": "Aracaju", "estado": "SE"}
        ],
        "vendas": [
            {"categoria": "Acessórios", "vendas_mensais": 90, "faturamento_mensal": 27000.0}
        ]
    },
    {
        "id_loja": 20,
        "nome_loja": "Loja Upsilon",
        "detalhes": [
            {"cidade": "Teresina", "estado": "PI"}
        ],
        "vendas": [
            {"categoria": "Livros", "vendas_mensais": 240, "faturamento_mensal": 72000.0}
        ]
    }
]
```
