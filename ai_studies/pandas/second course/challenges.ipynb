```python
import json
import requests
import pandas as pd
```

```python
dados_usuario = requests.get('https://jsonplaceholder.typicode.com/users')
```

```python
resultado = json.loads(dados_usuario.text)
```

<h3>Normalizing</h3>

```python
dados_usuarios_normalizados = pd.json_normalize(resultado, sep='_')
```

```python
dados_usuarios_normalizados
```

```python
dados_usuarios_normalizados.to_csv('dados_usuarios_normalizados.csv', sep=',')
```

```python
import pandas as pd
from sqlalchemy import create_engine, inspect, MetaData, Table, text
```

```python
engine = create_engine('sqlite:///:memory:')
url = 'https://raw.githubusercontent.com/alura-cursos/Pandas/main/clientes_banco.csv'
```

```python
clientes_csv = pd.read_csv(url)
```

```python
clientes_csv.to_sql('clientes2', engine, index=False)
```

```python
pd.read_sql_table('clientes2', engine).head()
```

```python
query = 'UPDATE clientes2 SET Rendimento_anual=300000 WHERE ID_Cliente=6840104'
with engine.connect() as conn:
    result = conn.execute(text(query))
    conn.commit()
    
```

```python
query = 'SELECT * FROM clientes2 WHERE ID_Cliente=6840104'
filtro_especifico = pd.read_sql(query, engine)
filtro_especifico
```
