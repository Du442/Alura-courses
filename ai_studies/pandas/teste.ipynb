<h2>Creating the DF</h2>

```python
import pandas as pd

values = {"Maça":[2.50], "Banana":[1.20], "Uva":[4.00]}
df = pd.DataFrame(values)
```

```python
df.T
```

```python
values_2 = {"Fruta":["Maça", "Banana", "Uva"],
            "Preço":[2.50, 1.20, 4.00]}
df_correta = pd.DataFrame(values_2)
```

```python
df_correta
```

<h2>Filtragem de dados</h2>

```python
df_correta['Preço'].min()
```

```python
df_correta[df_correta['Preço'] > 3.0]
```

<h2>Adição de elemento</h2>

```python
df_correta['Quantidade'] = [3, 5, 2]
```

```python
df_correta
```

<h2>Operações Matemáticas entre Colunas</h2>

```python
resultado_multiplicacao = df_correta['Preço'] * df_correta['Quantidade']
```

```python
df_correta['Total'] = resultado_multiplicacao
```

```python
df_correta
```

<h2>Ordenar os dados</h2>

```python
df_correta.sort_values(by='Total', ascending=False)
```

<h2>Resumos Estatísticos</h2>

```python
df_correta.describe()
```

```python
df_correta.info()
```

<h2>Lidar com Dados Ausentes</h2>

```python
df_correta.dropna()
```

```python
df_correta.fillna(0)
```

<h2>Agrupamento de Dados</h2>

```python
agrupamento_dados = df_correta.groupby(['Fruta'])[['Preço', 'Quantidade', 'Total']].sum().sort_values(by='Total', ascending=False)
```

```python
agrupamento_dados
```

```python
agrupamento_dados_preço = df_correta.groupby(['Fruta'])[['Preço']].sum().sort_values(by='Preço', ascending=False)
agrupamento_dados_preço
```

```python
df_correta.groupby(['Fruta', 'Preço'])
```

```python
df_correta.groupby(['Fruta', 'Preço'])[['Quantidade']].sum()
```

```python
df_correta.groupby('Fruta').agg({'Quantidade': 'sum', 'Preço': 'mean'}).sort_values(by='Preço', ascending=False).index
```

```python
df_correta.groupby('Fruta').agg({'Quantidade': 'sum', 'Preço': 'mean'}).sort_values(by='Preço', ascending=False).reset_index()
```
