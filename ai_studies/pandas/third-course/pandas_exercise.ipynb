<h3>Exercicio aula 3</h3>

```python
import pandas as pd

url = '1-SEEG10_GERAL-BR_UF_2022.10.27-FINAL-SITE.xlsx'

df = pd.read_excel(url, sheet_name='GEE Estados')
```

```python
colunas_info = list(df.loc[:, 'Nível 1 - Setor': 'Produto'].columns)
anos_info = list(df.loc[:, 1970:2021].columns)

new_df = df.melt(id_vars=colunas_info, value_vars=anos_info, var_name='Ano', value_name='Emissão')
```

```python
new_df
```

```python
nova_df_por_ano = new_df.pivot_table(aggfunc='mean', index='Ano', values='Emissão', columns='Nível 1 - Setor')
nova_df_por_ano
```

```python
nova_df_por_ano.plot(subplots=True ,figsize=(10,6))
```
