```python
dados = {
    'Nome': ['Jorge', 'Giuberto', 'Renan', 'Pedro'],
    'Turma': ['Terceiro', 'Terceiro', 'Terceiro', 'Segundo'],
    'Notas': [6.7, 8.5, 9.0, 7.5]
}

df = pd.DataFrame(dados)
```

```python
dados_brutos = {
    'Vendedor': ['Ana', 'Bruno', 'Ana', 'Carlos', 'Bruno', 'Ana', 'Carlos', 'Carlos'],
    'Produto': ['Celular', 'Fone', 'Celular', 'Tablet', 'Fone', 'Tablet', 'Celular', 'Tablet'],
    'Vendas_R$': [2500, 150, 2500, 1200, 150, 1100, 2500, 1300],
    'Data': ['2024-01-05', '2024-01-05', '2024-01-06', '2024-01-06', '2024-01-07', '2024-01-07', '2024-01-08', '2024-01-08'],
    'Comissao_%': [10, 5, 10, 8, 5, 8, 10, 8]
}

df_vendas = pd.DataFrame(dados_brutos)
df_vendas
```
